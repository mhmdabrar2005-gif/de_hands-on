# Ingesting from S3 volume

```python
#Define the path to the source volume
volume_path = "/Volumes/01_dev_bronze/source/source_volume/"

#Finding the names of the files in the source volume
files = dbutils.fs.ls(volume_path)
table_names_in_source=[]
for f in files:
    table_names_in_source.append(f.name)

#Loop through the files and write them to the raw database
for table_name in table_names_in_source:
  df = spark.read.format("csv").load("/Volumes/01_dev_bronze/source/source_volume/"+table_name,header=True)
  df.write.option("overwrite", "true").saveAsTable("`01_dev_bronze`.`raw`.`"+table_name.split(".")[0]+"`",mode="overwrite")
```
